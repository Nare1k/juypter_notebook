```python
# Exercise 2 

# Gets the users input for the fibonacci sequence
number = input(int("How many numbers of Fibonacci Sequence would you like?"))

def fibonacci(number):
    fib_sequence = [0, 1]
    while len(fib_sequence) < number: # While the length of the fibonacci sequence is less than the input number it continue.
        next_fib = fib_sequence[-1] + fib_sequence[-2]
        fib_sequence.append(next_fib)
    return fib_sequence
# Calls the fibonacci function and prints the result.
fib_sequence = fibonacci(number)
print(fib_sequence)

```

```python
# Exercise 5

def find_min_max(numbers):
    if not numbers:
        raise ValueError("The input iterable cannot be empty.")
    return min(numbers), max(numbers)

# Example usage:
numbers = [4, 2, 9, 1, 5, 6]
min_val, max_val = find_min_max(numbers)

print(f"Min value is: {min_val}")
print(f"Max value is: {max_val}")
```

```python
# Exercise 8 
import math

# User input
x = float(input("Enter the value of x: "))
N = int(input("Enter the value of N: "))

def series_sum(x, N):
    total = 0
    for n in range(1, N + 1):
        numerator = x * (5 ** (2 * n))
        denominator = n + math.factorial(n + 1)
        total += numerator / denominator
    return total


result = series_sum(x, N)
print(f"The sum of the series is: {result}")
```

```python
# Exercise 9
import numpy as np
import matplotlib.pyplot as plt
import cv2

# Load the image
image_path = "C:\Users\Luke Family\Pictures\matt.jpg"  # Replace with your image path
image = cv2.imread(image_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Convert BGR to RGB

# Generate reflections
horizontal_flip = np.fliplr(image)  # Horizontal reflection
vertical_flip = np.flipud(image)    # Vertical reflection

# Display images
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(image)
axes[0].set_title("Original Image")
axes[0].axis("off")

axes[1].imshow(horizontal_flip)
axes[1].set_title("Horizontally Mirrored")
axes[1].axis("off")

axes[2].imshow(vertical_flip)
axes[2].set_title("Vertically Mirrored")
axes[2].axis("off")

plt.show()
```
